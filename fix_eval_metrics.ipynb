```python
## Helper functions
def calc_eval(row, ks = [3,5,10]):
    results = {}
    for k in ks:
        results[f'precision_{k}'] = calculate_precision_at_k(row['retrieved_functions'], row['edit_functions'], k=k)
        results[f'recall_{k}'] = calculate_recall_at_k(row['retrieved_functions'], row['edit_functions'], k=k)
        results[f'f1_{k}'] = calculate_f1_at_k(row['retrieved_functions'], row['edit_functions'], k=k)
        results[f'iou_{k}'] = calculate_iou(row['retrieved_functions'], row['edit_functions'], k=k)
    results['mrr'] = calculate_rr(row['retrieved_functions'], row['edit_functions'])
    return pd.Series(results)

def process_and_reupload_eval_results(client,run_id: str):
    """
    Downloads evaluation_results.csv for a given MLflow run,
    loads it into a pandas DataFrame, allows for modification,
    reuploads it, and logs corrected metrics.
    """
    
    # Step 1: Download artifact
    local_dir = tempfile.mkdtemp()
    local_path = client.download_artifacts(run_id, "evaluation_results.csv", local_dir)
    
    # Step 2: Load into DataFrame
    df = pd.read_csv(local_path)
    
    # Step 2.5: Recalculate metrics
    df['retrieved_functions'] = df['retrieved_functions'].apply(literal_eval)
    df['edit_functions'] = df['edit_functions'].apply(literal_eval)
    metric_columns = [
    'precision_3', 'precision_5', 'precision_10',
    'recall_3', 'recall_5', 'recall_10',
    'f1_3', 'f1_5', 'f1_10',
    'iou_3', 'iou_5', 'iou_10',
    'mrr'
    ]

    recalculated_metrics = df.apply(calc_eval, axis=1)
    df[metric_columns] = recalculated_metrics[metric_columns]
    
    # Step 3: Save modified DataFrame to a new CSV
    corrected_path = os.path.join(local_dir, "evaluation_results_corrected.csv")
    df.to_csv(corrected_path, index=False)
    
    # Step 4: Upload corrected CSV back to run
    client.log_artifact(run_id, corrected_path, artifact_path="evaluation")
    
    # Step 5: Log corrected metrics (to appear in MLflow UI graphs)
    mean_metrics = df[metric_columns].mean()
    for metric_name, metric_value in mean_metrics.items():
        client.log_metric(run_id, metric_name, metric_value)
    
    return df
```

```python
client = MlflowClient(tracking_uri="http://127.0.0.1:5000")
runs_df = client.search_runs(experiment_ids=['1']) 
for run in runs_df:
    run_id = run.info.run_id
    if run.info.status != 'FINISHED':
        print(f"Skipping run {run_id} as it is not finished.")
        continue
    _ = process_and_reupload_eval_results(client=client,run_id=run_id)
```
