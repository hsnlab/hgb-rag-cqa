```python
from src.utils.qdrant_store import QdrantStore
import json

# Initialize QdrantStore with Neo4j connection details
qdrant_config_path = "_/qdrant_config.json"
try:
    with open(qdrant_config_path, "r") as f:
        qdrant_config = json.load(f)
except FileNotFoundError:
    print(f"Error: Qdrant config file not found at {qdrant_config_path}")
except json.JSONDecodeError:
    print(f"Error: Qdrant config file is not valid JSON: {qdrant_config_path}")

neo4j_config_path = "_/neo4j_config.json"
try:
    with open(neo4j_config_path, "r") as f:
        neo4j_config = json.load(f)
except FileNotFoundError:
    print(f"Error: Neo4j config file not found at {neo4j_config_path}")
except json.JSONDecodeError:
    print(f"Error: Neo4j config file is not valid JSON: {neo4j_config_path}")

store = QdrantStore(
            model_name=qdrant_config["model_name"],
            qdrant_url=qdrant_config["url"],
            collection_name=qdrant_config["collection"],
            api_key=qdrant_config["api_key"],
            distance_type=qdrant_config["distance"],
            neo4j_uri=neo4j_config["url"],
            neo4j_auth=(neo4j_config["user"],neo4j_config["password"])
        )
```

```python
# Delete existing collection if it exists
store.clear_collection()
```

```python
# Add issues from Neo4j to Qdrant
store.add_from_neo4j(node_label="ISSUE", text_property="issue_title",id_property="ID", metadata_type="issue_title")
print(store.get_collection_info())
store.add_from_neo4j(node_label="ISSUE", text_property="issue_body",id_property="ID", metadata_type="issue_body")
print(store.get_collection_info())
#store.add_from_neo4j(node_label="ISSUE", text_property="issue_labels",id_property="ID", metadata_type="issue_label")
```

```python
# Add functions from Neo4j to Qdrant
store.add_from_neo4j(node_label="FUNCTION", text_property="function_code",id_property="ID", metadata_type="function_code")
print(store.get_collection_info())
store.add_from_neo4j(node_label="FUNCTION", text_property="combinedName",id_property="ID", metadata_type="function_name")
print(store.get_collection_info())
#store.add_from_neo4j(node_label="FUNCTION", text_property="docstring",id_property="ID", metadata_type="function_docstring")
#print(store.get_collection_info())
```

```python
# Add PRs from Neo4j to Qdrant
store.add_from_neo4j(node_label="PR", text_property="title",id_property="ID", metadata_type="pr_title")
print(store.get_collection_info())
store.add_from_neo4j(node_label="PR", text_property="body",id_property="ID", metadata_type="pr_body")
print(store.get_collection_info())
```

```python
# Add imports from Neo4j to Qdrant
store.add_from_neo4j(node_label="IMPORT", text_property="import_name",id_property="ID", metadata_type="import_name")
```

```python
# Add classes from Neo4j to Qdrant
store.add_from_neo4j(node_label="CLASS", text_property="name",id_property="ID", metadata_type="class_name")
```

```python
# Add clusters from Neo4j to Qdrant
store.add_from_neo4j(node_label="CLUSTER", text_property="summary",id_property="ID", metadata_type="semantic_cluster")
```
